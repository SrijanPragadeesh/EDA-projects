```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df=pd.read_csv("/content/Ecommerce_Order_Test_Dataset.csv")
df.head()
```

```python
print("number of row",df.shape[0])
print("number of column",df.shape[1])
```

```python
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
df['City'] = df['City'].fillna('Unknown')
df['Customer_Name'] = df['Customer_Name'].fillna('Unknown')
print()
```

```python
mean_rating = df['Rating'].mean()

df['Rating'] = df['Rating'].fillna(mean_rating)
```

```python
df['Payment_Mode'] = df['Payment_Mode'].fillna('Unknown')
```

```python
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates()

print("Duplicate rows removed.")
```

```python
print("New shape of DataFrame:", df.shape)
```

```python
df['Gender'] = df['Gender'].astype(str).str.strip().str.title()
df['City'] = df['City'].astype(str).str.strip().str.title()
df['Payment_Mode'] = df['Payment_Mode'].astype(str).str.strip().str.title()
```

```python
print(df['Gender'].unique())
print(df['City'].unique())
print(df['Payment_Mode'].unique())
```

```python
df['Total_Price'] = df['Quantity'] * df['Unit_Price']
df['Discount_Amount'] = df['Total_Price'] * df['Discount']
df['Final_Amount'] = df['Total_Price'] - df['Discount_Amount']
```

```python
display(df[
    ['Order_ID',
     'Product',
     'Quantity',
     'Total_Price',
     'Discount_Amount',
     'Final_Amount']
])
```

```python
total_quantity = df['Quantity'].sum()
average_unit_price = df['Unit_Price'].mean()
maximum_unit_price = df['Unit_Price'].max()
minimum_unit_price = df['Unit_Price'].min()
average_rating = df['Rating'].mean()

print("Total Quantity Ordered:", total_quantity)
print("Average Unit Price:", average_unit_price)
print("Maximum Unit Price:", maximum_unit_price)
print("Minimum Unit Price:", minimum_unit_price)
print("Average Rating:", average_rating)
```
